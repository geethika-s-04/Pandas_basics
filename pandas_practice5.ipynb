```python
import pandas as pd
df=pd.read_csv(r"C:\Users\geeth\OneDrive\Desktop\Data analysis\employee_data.csv")
print(df)
```

```python
result=df.groupby("Department")["Salary"].agg(["mean","count"])
print(result)
```

```python
sort=df.sort_values("Joining_Date")
print(sort.iloc[0])
```

```python
sal_range=[]
for i in df["Salary"]:
    if i<=50000:
        sal_range.append("0-50000")
    elif i<=60000:
        sal_range.append("50001-60000")
    elif i<=70000:
        sal_range.append("60001-70000")
    else:
        sal_range.append("70001+")
df["Salary_Range"]=sal_range
print(df["Salary_Range"].value_counts())
```

```python
df1=pd.read_csv(r"C:\Users\geeth\OneDrive\Desktop\Data analysis\exam_scores.csv")
print(df1)
```

```python
result=df1[["MathScore","EnglishScore","ScienceScore"]].agg(["max","min","mean","median"])
print(result)
```

```python
df1["Average"]=(df1["MathScore"]+df1["EnglishScore"]+df1["ScienceScore"]).div(3)
print(df1.loc[df1["Average"].idxmax()])
```

```python
math=(df1["MathScore"]<50).sum()
eng=(df1["EnglishScore"]<50).sum()
sci=(df1["ScienceScore"]<50).sum()
print("Count of students who scored below 50 in each subject:")
print("Math:",math)
print("English",eng)
print("Science:",sci)
print("Maximum Value:")
maxim=max(math,eng,sci)
if maxim==math:
    print("Maths:",math)
if maxim==eng:
    print("English:",eng)
if maxim==sci:
    print("Science:",sci)
```
