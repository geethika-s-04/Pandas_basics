```python
import pandas as pd
import numpy as np
df=pd.read_csv(r"C:\Users\geeth\OneDrive\Desktop\Data analysis\student_data.csv")
print(df)
```

```python
arr=df.to_numpy()
print(arr)
```

```python
print(np.mean(arr[:,1]))
```

```python
std_val=np.std(arr[:,4])
print(std_val)
```

```python
median_val=np.median(arr[:,2])
print(median_val)
```

```python
filtered_val=[]
for i in arr:
    if i[1]>90 or i[2]>90 or i[3]>90:
        filtered_val.append(True)
    else:
        filtered_val.append(False)
print(filtered_val)
print("==================================")
newarr=arr[filtered_val]
print(newarr)
```

```python
filtered_val=[]
for i in arr:
    if i[4]<75:
        filtered_val.append(True)
    else:
        filtered_val.append(False)
print(filtered_val)
print("==================================")
newarr=arr[filtered_val]
print(newarr)
```

```python
avg_math=np.mean(arr[:,1])
avg_sci=np.mean(arr[:,2])
avg_eng=np.mean(arr[:,3])
filtered_val=[]
for i in arr:
    if i[1]>avg_math and i[2]>avg_sci and i[3]>avg_eng:
        filtered_val.append(True)
    else:
        filtered_val.append(False)
print(filtered_val)
print("==================================")
newarr=arr[filtered_val]
print(newarr)
```

```python
total=arr[:,1]+arr[:,2]+arr[:,3]
print(total)
```

```python
high=total[0]
student=arr[0,0]
for i in range(len(total)):
    if total[i]>high:
        high=total[i]
        student=arr[i,0]
print("Student:",student)
print("highest total:",high)
```

```python
filtered_val=[]
for i in arr:
    if i[1]>=50 and i[2]>=50 and i[3]>=50 and i[4]>=75:
        filtered_val.append("Pass")
    else:
        filtered_val.append("Fail")
print(filtered_val)

df["Pass/Fail"]=filtered_val
print(df)
```

```python
percent=(total/300)*100
print(percent)
df["Percentage"]=percent
print(df)
```

```python
grade=[]
for i in percent:
    if i>=85:
        grade.append("A")
    elif i>=70:
        grade.append("B")
    elif i>=50:
        grade.append("C")
    else:
        grade.append("F")
print(grade)
df["Grade"]=grade
print(df)
```
