```python
designation=[]
experience=[]
for i in df["Department"]:
    if i=="IT" or i=="it":
        designation.append("Software Engineer")
        experience.append(4)
    elif i=="HR" or i== "Human Resource":
        designation.append("HR Executive")
        experience.append(3)
    elif i=="Marketing":
        designation.append("Marketing Manager")
        experience.append(5)
    elif i=="Finance":
        designation.append("Accountant")
        experience.append(6)
    elif i=="Sales":
        designation.append("Sales Executive")
        experience.append(4)
    else:
        designation.append("Support Engineer")
        experience.append(3)

df1=pd.DataFrame({"Designation":designation,"Experience":experience})
df=df.join(df1)
print(df)
```
