```python
print(df.dtypes)
print(df.isnull().sum())
```

```python
print(df["IMDb_Rating"].agg(["mean","min","max"]))
```

```python
print(df["Box_Office"].mean())
print("---------------------------------")
print(df.loc[df["Box_Office"].idxmax()])
print("----------------------------------")
print(df.loc[df["Box_Office"].idxmin()])
```

```python
result=df.groupby("Genre")["Duration_min"].mean()
print(result)
print("----------------------------------")
print(result.idxmax())
```
