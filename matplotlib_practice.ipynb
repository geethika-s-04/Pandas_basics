```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv(r"C:\Users\geeth\Downloads\matplotlib_pandas_practice_dataset.csv")
print(df.head())
```

```python
print(df.duplicated().sum())
df=df.drop_duplicates()
print(df.shape)
```

```python
print(df.isnull().sum())
```

```python
print(df["Age"].dtype)
print(df["Salary"].dtype)
```

```python
df["Age"]=pd.to_numeric(df["Age"],errors='coerce')
df["Salary"]=pd.to_numeric(df["Salary"],errors='coerce')
print(df["Age"].dtype)
print(df["Salary"].dtype)
```

```python
df["Age"]=df["Age"].fillna(df["Age"].median())
df["Salary"]=df["Salary"].fillna(df["Salary"].median())
print(df.isnull().sum())
```

```python
df=df[(df["Age"]>=18)&(df["Age"]<=100)]
df=df[df["Salary"]>0]
print(df)
```

```python
print(df["Department"].unique())
print(df["City"].unique())
```

```python
df["Department"]=df["Department"].replace("it","IT")
df["City"]=df["City"].replace("Mumbi","Mumbai")
df=df.dropna()
print(df["Department"].unique())
print(df["City"].unique())
```

```python
df["Total_Income"]=df["Salary"]+df["Sales"]
result=df[df["Total_Income"]>100000]
print(result)
```

```python
x=df["EmployeeID"]
y=df["Salary"]
plt.plot(x,y,marker='o',mfc='black',color='red')
plt.title("EmployeeID vs Salary")
plt.xlabel("EmployeeID")
plt.ylabel("Salary")
plt.grid(True)
plt.show()
```

```python
x=df["EmployeeID"]
y=df["Sales"]
plt.plot(x,y,marker='o',ms=12,mfc='blue',mec='red',linestyle='')
plt.title("EmployeeID vs Sales")
plt.xlabel("EmployeeID")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
x=df["EmployeeID"]
plt.plot(x,df["Salary"],marker='*',mfc='green',mec='blue',color='red')
plt.plot(x,df["Sales"],marker='s',mfc='y',mec='m',color='black')
plt.title("EmployeeID vs Salary and Sales")
plt.xlabel("EmployeeID")
plt.ylabel("Salary and Sales")
plt.grid(True)
plt.show()
```
