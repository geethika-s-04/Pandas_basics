```python
import pandas as pd
df=pd.read_csv(r"C:\Users\geeth\Downloads\pandas_practice_groupby_to_convertdtypes_dataset.csv")
print(df)
```

```python
print(df.shape)
print(df.select_dtypes(include="number"))
```

```python
print("****Average****")
print(df.groupby("Department")["Salary"].mean())
print("*****Max and Min*****")
print(df.groupby("Department")["Bonus"].agg(["max","min"]))
```

```python
print(df.sort_values(["Department","Salary"],ascending=[True,False]))
```

```python
print(df.take([1,5,10,15]))
```

```python
print(df.truncate(before=10,after=20))
```

```python
df.eval("TotalIncome= Salary + Bonus",inplace=True)
print(df.sort_values("TotalIncome",ascending=False))
```

```python
temp=df.copy()
temp.loc[temp["Salary"]<=60000,"Salary"]=0
print(temp)
```

```python
df.eval("Updated_Bonus= Bonus+(Bonus*0.10)",inplace=True)
print(df)
```

```python
group=df.groupby("Department")["Salary"].sum()
group=group.sort_values(ascending=False)
print(group)
```

```python
result=df[df["Department"].isin(["IT","Finance"])]
print(result.select_dtypes(include="number"))
```

```python
new_df=pd.DataFrame({"Salary":[70000,72000,68000,75000,71000]},index=[0,1,2,3,4])
df.update(new_df)
print(df)
```

```python
df["Age"]=df["Age"].replace("Thirty","30")
print(df["Age"])
```

```python
df["Age"]=df["Age"].astype("int")
df["Salary"]=df["Salary"].astype("int")
df["Bonus"]=df["Bonus"].astype("int")
df=df.convert_dtypes()
print(df.dtypes)
```

```python
print(df.values)
print("------------------------")
print(df.take([0,1,2,3,4]))
```

```python
df=df.assign(AnnualSalary=df["Salary"].mul(12))
print(df.sort_values("AnnualSalary",ascending=False).head())
```

```python
result=df.groupby("Department")[["Experience","Salary"]].mean().sort_values("Experience")
print(result)
```

```python
trunc=df.truncate(before=15,after=30)
print(trunc.sort_values("Experience",ascending=False))
```

```python
print(df.select_dtypes(include="object"))
df=df.convert_dtypes()
print(df.dtypes)
```
